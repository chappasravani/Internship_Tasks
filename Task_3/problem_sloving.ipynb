1.Pseudocode: Word Frequency

1. Start.
2. Take a sentence as input.
3. Split the sentence into individual words.
4. Create an empty dictionary called frequency
5. For each word in the sentence:
   - If the word already exists in the dictionary, increase its count by 1.
   - Otherwise, add the word to the dictionary with a count of 1.
6. Return the dictionary.
7. Display the result.
8. Stop.

2.Pseudocode: Character Frequency

1. Start.
2. Take a word as input.
3. Create an empty dictionary called frequency.
4. Traverse each character in the word.
5. If the character already exists in the dictionary, increase its count by 1.
6. Otherwise, add the character to the dictionary with a count of 1.
7. Return the dictionary.
8. Display the result.
9. Stop.



3.Pseudocode:non-repeating-character

1. Start.
2. Take a word as input.
3. Create an empty dictionary to store the frequency of each character.
4. Traverse each character in the word.
5. If the character already exists in the dictionary, increase its count by 1.
6. Otherwise, add the character to the dictionary with a count of 1.
7. Traverse the word again from left to right.
8. Check the frequency of each character.
9. If a character's frequency is 1, return that character and stop searching.
10. If no character has a frequency of 1, return None.
11. Display the result.
12. Stop.



4.Pseudocode:Two Words Are Anagrams

1. Start.
2. Take two words as input.
3. Sort the characters of the first word alphabetically.
4. Sort the characters of the second word alphabetically.
5. Compare the two sorted words.
6. If both sorted words are equal, return True.
7. Otherwise, return False.
8. Display the result.
9. Stop.

5.Pseudocode:First Repeating Character

1. Start.
2. Take a word as input.
3. Create an empty dictionary to store characters that have already appeared.
4. Traverse each character in the word from left to right.
5. Check whether the character already exists in the dictionary.
6. If it exists, return that character and stop searching.
7. Otherwise, add the character to the dictionary.
8. If no repeating character is found, return none.
9. Display the result.
10. Stop.

6.Pseudocode:All Zeros to the End

1. Start.
2. Take a list of numbers as input.
3. Create an empty list called result.
4. Traverse the input list.
5. Add every non-zero number to result.
6. Traverse the input list again.
7. Add every zero to result.
8. Return the result list.
9. Display the result.
10. Stop.

7.Pseudocode: Second-Highest Distinct Number

1. Start.
2. Take a list of numbers as input.
3. Remove duplicate numbers from the list.
4. Sort the unique numbers in ascending order.
5. Check whether at least two unique numbers exist.
6. If fewer than two unique numbers exist, return none.
7. Otherwise, return the second-last number in the sorted list.
8. Display the result.
9. Stop.

8.Pseudocode:Remove Duplicates from a List

1. Start.
2. Take a list of numbers as input.
3. Create an empty list called result.
4. Traverse each number in the input list.
5. Check whether the number already exists in result.
6. If it does not exist, add the number to result.
7. If it already exists, skip it.
8. Return the result list.
9. Display the result.
10. Stop.

9.Pseudocode:Missing Number from 1 to N

1. Start.
2. Take a list of numbers and the value of N as input.
3. Traverse the numbers from 1 to N.
4. Check whether each number exists in the given list.
5. If a number does not exist, return that number.
6. If no number is missing, return none.
7. Display the result.
8. Stop.

10.Pseudocode:irst Pair Whose Sum Equals the Target

1. Start.
2. Take a list of numbers and a target value as input.
3. Traverse each number from left to right using an outer loop.
4. For each number, check every number that comes after it using an inner loop.
5. Add the two numbers and compare their sum with the target.
6. If the sum equals the target, return the pair immediately.
7. If no matching pair is found, return none.
8. Display the result.
9. Stop.

11.Pseudocode:Number Closest to Zero

1. Start.
2. Take a list of numbers as input.
3. If the list is empty, return none.
4. Set the first number as the closest number.
5. Traverse each number in the list.
6. Compare the absolute value of the current number with the absolute value of the closest number.
7. If the current number is closer to zero, update the closest number.
8. If both numbers are equally close to zero, choose the positive number.
9. Return the closest number.
10. Display the result.
11. Stop.

12.Pseudocode:Common Elements in Two Lists

1. Start.
2. Take two lists as input.
3. Create an empty list called result.
4. Traverse each element in the first list.
5. Check whether the element exists in the second list.
6. Also check whether the element is already present in result.
7. If it exists in the second list and is not already in result, add it.
8. Repeat until all elements in the first list are checked.
9. Return the result list.
10. Display the result.
11. Stop.

13.Pseudocode:Rotate a List to the Right by K Positions

1. Start.
2. Take a list of numbers and the value of k as input.
3. If the list is empty, return an empty list.
4. Find the remainder when k is divided by the length of the list.
5. If the remainder is zero, return the original list.
6. Otherwise, separate the last k elements from the list.
7. Place those elements at the beginning of the list.
8. Place the remaining elements after them.
9. Return the rotated list.
10. Display the result.
11. Stop.

14.Pseudocode:Maximum Sum of Adjacent Elements

1. Start.
2. Take a list of numbers as input.
3. Check whether the list contains at least two elements.
4. If it contains fewer than two elements, return none.
5. Calculate the sum of the first two adjacent elements and store it as the maximum sum.
6. Traverse the remaining adjacent pairs in the list.
7. Calculate the sum of each adjacent pair.
8. If the current sum is greater than the maximum sum, update the maximum sum.
9. After checking all adjacent pairs, return the maximum sum.
10. Display the result.
11. Stop.

15.Pseudocode:All Unique Pairs Whose Sum Equals the Target

1. Start.
2. Take a list of numbers and a target value as input.
3. Create an empty list called result.
4. Use an outer loop to select each element from the list.
5. Use an inner loop to select each element that comes after it.
6. Add the two selected elements.
7. If their sum equals the target, arrange the pair in ascending order.
8. Check whether the pair is already present in result.
9. If the pair is not present, add it to result.
10. After checking all pairs, sort result by the first element of each pair.
11. Return the result.
12. Display the result.
13. Stop.
